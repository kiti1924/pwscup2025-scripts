```python
import pandas as pd
from pathlib import Path

# ディレクトリと出力ファイル
indir = Path(r"out_attack")
outdir = Path(r"submission")
outfile = outdir / "FF22.csv"

# f"out_attack/{mode_anon}{{id:02d}}_inferred_hungarian_auto_k300_weighted_entropy_{alpha}.csv",
# Fij_01.csv ～ Fij_20.csv のパス (excluding 21 and 22)
files = [indir / f"CC{i:02d}_inferred_hungarian_auto_k300_weighted_entropy_0.13.csv" 
         for i in range(1, 25)]  # Changed to exclude both 21 and 22

# Create output directory if it doesn't exist
outdir.mkdir(parents=True, exist_ok=True)

# 各ファイルを読み込んで列方向に結合
dfs = [pd.read_csv(f, header=None) for f in files]
merged = pd.concat(dfs, axis=1)

# 保存（ヘッダもインデックスも出力しない）
merged.to_csv(outfile, index=False, header=False)

print(f"Saved {outfile}, shape={merged.shape}")
```
