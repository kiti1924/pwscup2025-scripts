```python
import pandas as pd
import os
import numpy as np
```

```python
df1 = pd.DataFrame(columns=["anonymization", "stats_diff", "LR_asthma_diff", "KW_IND_diff", "Ci_utility"])
```

```python
# testing for team 22 only
def get_teams():
    arr = np.arange(22, 23)
    return arr  
```

```python
files = os.listdir("out_anonymized")
for file in files:
    result = []
    if file.startswith("CC22_1") and file.endswith("_eval.txt"):
        # print(file)
        result.append(file)
        with open(f"out_anonymized/{file}") as f:
            lines = f.readlines()
            for line in lines:
                if ":" in line:
                    score_a = line.split(":")[1].strip()
                    score_b = score_a.split()[0]
                    result.append(float(score_b))
        # print(result)
        df1 = pd.concat([df1, pd.DataFrame([result], columns=df1.columns)], ignore_index=True)
```
